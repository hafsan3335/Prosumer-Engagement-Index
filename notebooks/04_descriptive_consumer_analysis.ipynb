# 04. Descriptive Consumer & Provider Participation Analysis
**Research Objective**: Analyze weighted and unweighted adoption rates, frequency distributions, and multi-sector participation breadth across collaborative platforms.


```python
import pandas as pd
from src.config import PROC_FULL_CSV

df = pd.read_csv(PROC_FULL_CSV)
users = df[df['is_active_user'] == 1]

print("Top Consumer Sectors:")
cons_sectors = [c for c in df.columns if c.startswith('cons_sector_')]
print((df[cons_sectors].mean() * 100).round(2).sort_values(ascending=False))

print("
Top Provider Sectors:")
prov_sectors = [c for c in df.columns if c.startswith('prov_sector_')]
print((df[prov_sectors].mean() * 100).round(2).sort_values(ascending=False))
```

### Consumer Behaviour Interpretation
- **Transport** (car sharing, ride hailing) and **Accommodation** (apartment/home sharing) represent the primary anchor sectors of the collaborative economy, followed by food delivery and household services.
- While consumption is widespread across multiple sectors, provider activity is concentrated primarily in accommodation and transport, reflecting high asset-intensity monetization.
