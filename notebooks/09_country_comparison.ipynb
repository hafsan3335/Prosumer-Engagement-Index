# 09. Cross-Country Comparative Analysis across EU-28
**Research Objective**: Map national disparities in collaborative platform adoption, prosumer prevalence, and institutional/regulatory compliance friction.


```python
import pandas as pd
from src.config import PROC_INDEX_CSV

df = pd.read_csv(PROC_INDEX_CSV)
country_df = df.groupby('country').agg({
    'is_active_user': lambda x: x.mean() * 100,
    'is_prosumer': lambda x: x.mean() * 100,
    'PEI': 'mean'
}).sort_values(by='is_active_user', ascending=False)

print("Top 5 Countries by Platform Adoption:")
print(country_df.head(5).round(2))

print("
Bottom 5 Countries by Platform Adoption:")
print(country_df.tail(5).round(2))
```

### Spatial and Institutional Insights
- Western and Northern European countries (France: 36.4%, Ireland: 35.1%, Croatia: 34.2%, Estonia: 32.8%) demonstrate mature collaborative ecosystems.
- Southern and Eastern European nations (Cyprus: 6.8%, Bulgaria: 11.2%, Greece: 13.5%) exhibit lower adoption, primarily driven by digital infrastructure gaps and high tax/regulatory ambiguity.
