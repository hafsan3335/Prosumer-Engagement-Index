### Interpretation of Survey Architecture
- **Skip Logic 1 (Consumer Questions)**: Questions `q2` (sectors used), `q3` (substitution), `q4` (perceived advantages), `q5` (perceived disadvantages), and `q6` (recommendation) were asked exclusively to respondents who indicated previous platform consumption (`d8 >= 2`). Non-consumers skipped these questions structurally.
- **Skip Logic 2 (Provider Questions)**: Questions `q10` (sectors offered), `q11` (provider motivations), and `q12` (provider barriers) were asked strictly to respondents offering services (`d9 >= 2`).
- **Analytical Rule**: Structural skips are recoded to 0 (indicating absence of activity/motive) rather than imputed as missing values.
