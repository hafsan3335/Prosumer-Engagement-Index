```python
import pandas as pd
from src.config import PROC_SEGMENTS_CSV

seg_df = pd.read_csv(PROC_SEGMENTS_CSV)
print("Sustainability Motivation by Segment (%):")
print((seg_df.groupby('segment_name')['prov_mot_sustainability'].mean() * 100).round(2))

print("
Recommendation Advocacy by Segment (%):")
print((seg_df.groupby('segment_name')['recommend_high'].mean() * 100).round(2))
```

- Sustainability motives are heavily concentrated in **Persona 3 (Sustainability Co-Creators)**, who recirculate idle household goods and peer services.
- Micro-providers and sustainability co-creators exhibit the highest advocacy rates, representing optimal cohorts for platform referral and ambassador initiatives.
