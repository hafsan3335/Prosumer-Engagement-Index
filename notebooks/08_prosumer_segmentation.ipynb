# 08. Consumer & Prosumer Behavioral Segmentation
**Research Objective**: Apply K-Means clustering to active participants, evaluate silhouette diagnostics across cluster counts, and construct actionable marketing persona profiles.


```python
import pandas as pd
from src.segmentation import prepare_clustering_features, evaluate_cluster_solutions, fit_final_segmentation
from src.config import PROC_INDEX_CSV

idf = pd.read_csv(PROC_INDEX_CSV)
users_df, feature_cols, X_scaled = prepare_clustering_features(idf)

eval_df = evaluate_cluster_solutions(X_scaled)
print("Cluster Solution Evaluation Metrics:")
print(eval_df)

segmented_users, profile_table = fit_final_segmentation(users_df, X_scaled, k=4)
print("
Segment Distribution:")
print(segmented_users['segment_name'].value_counts(normalize=True).round(3) * 100)
```

### Segmentation Insights
- Four distinct personas emerge:
  1. *Occasional Frugal Consumers* (38.4%): Price-sensitive, single-sector, zero supply-side participation.
  2. *Convenience-Driven Urbanites* (32.8%): Digital ease, frequent transport/accommodation users.
  3. *Sustainability & Community Co-Creators* (14.6%): Dual prosumers motivated by asset efficiency and waste reduction.
  4. *Entrepreneurial Micro-Providers* (14.2%): Supply-side income earners and brand advocates.
