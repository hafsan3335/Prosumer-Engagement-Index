# 05. Construction of the Prosumer Engagement Index (PEI)
**Research Objective**: Formulate the multi-component additive Prosumer Engagement Index (PEI, 0–100) combining consumer frequency, provider frequency, activity breadth, and traditional channel substitution.


```python
import pandas as pd
from src.index import compute_pei
from src.config import PROC_FULL_CSV

df = pd.read_csv(PROC_FULL_CSV)
idf = compute_pei(df)

users = idf[idf['is_active_user'] == 1]
print("PEI Summary Statistics (Active Users N=5,872):")
print(users['PEI'].describe())

print("
Mean PEI by Profile Role:")
print(users.groupby('user_profile')['PEI'].agg(['count', 'mean', 'median', 'std']))
```

### Index Distribution Insights
- Active collaborative platform users achieve a mean PEI of **$18.2 / 100$** (Median = $16.7$).
- Dual-role **Prosumers** achieve a mean PEI of **$37.4 / 100$** (Median = $37.5$, Std = $13.5$), compared to **$13.4 / 100$** for consumer-only participants.
- The index demonstrates excellent discriminatory power across behavioral role depth.
